## 上午：重做本周卡住的 3 道练习
不查资料、不查记录，独立写出来才算过关。候选：

- 每列标准化
- 余弦相似度矩阵
- CSVReader

```python
# 重做开始
import numpy as np

# 固定随机种子，保证每次运行结果一致
np.random.seed(42)

print("=" * 50)
print("任务 1：词频统计前 10（sorted + lambda 法）")
print("=" * 50)
text = 'the quick brown fox jumps over the lazy dog the fox runs and the dog sleeps'
words = text.split()

d = {}
for w in words:
    d[w] = d.get(w, 0) + 1

# 按次数降序（-x[1]），次数相同时按单词升序（x[0]）
top10_sorted = sorted(d.items(), key=lambda x: (-x[1], x[0]))[:10]
print("Top 10 词频:", top10_sorted)


print("\n" + "=" * 50)
print("任务 2：每列标准化（NumPy 广播法）")
print("=" * 50)
# 造 100 行 3 列数据（模拟特征矩阵）
X = np.random.rand(100, 3) * 10 + 5
print("标准化前 - 前 2 行数据:\n", np.round(X[:2], 2))
print("标准化前 - 每列均值:", np.round(X.mean(axis=0), 4))

# (100, 3) 与 (3,) 直接广播，(3,) 代表每一列的均值/标准差
X_norm = (X - X.mean(axis=0)) / X.std(axis=0)

print("\n标准化后 - 前 2 行数据:\n", np.round(X_norm[:2], 2))
print("标准化后 - 每列均值:", np.round(X_norm.mean(axis=0), 4))  # 应该全是 0
print("标准化后 - 每列标准差:", np.round(X_norm.std(axis=0), 4))  # 应该全是 1


print("\n" + "=" * 50)
print("任务 3：余弦相似度矩阵（归一化点积法）")
print("=" * 50)
# 5 个 8 维随机向量
V = np.random.randn(5, 8)

# 1. 求每个向量的 L2 范数
norms = np.linalg.norm(V, axis=1)

# 2. 每行除以自己的范数，实现归一化（norms[:, None] 变 (5,1) 用于广播）
V_norm = V / norms[:, None] 

# 3. 归一化后的矩阵乘以自身的转置，直接得到余弦相似度
sim2 = V_norm @ V_norm.T

print("余弦相似度矩阵 (保留3位小数):")
print(np.round(sim2, 3))

# 4. 找每个向量最相似的索引（排除自己）
sim_copy = sim2.copy()  # 复制一份，避免污染原矩阵
np.fill_diagonal(sim_copy, -np.inf)  # 把对角线换成负无穷，这样绝对不会选到自己
most_similar_idx = np.argmax(sim_copy, axis=1)

print("\n每个向量的最相似向量编号:", most_similar_idx)
```
