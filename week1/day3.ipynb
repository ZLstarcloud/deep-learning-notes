> 类比：类 = 数据 + 行为，和你用过的语言一样。重点记 Python 写法：self、双下划线方法。

```python
class Counter:
    def __init__(self, start=0):
        self.count = start
    def add(self, n=1):
        self.count += n
        return self
    def __repr__(self):
        return f'Counter({self.count})'
c = Counter(5).add().add(3)
print(c)
```

## 4. 下午项目：CSVReader 类

```python
# 要求：
# 1. __init__(path) 读文件：第一行是表头，其余是数据（提示 str.split(',')）
# 2. column(name) -> 返回该列所有值的列表
# 3. mean(name) -> 返回该数值列的平均值
# 4. 文件不存在给出友好报错；列名写错抛 KeyError
# 先造一个 sales.csv（date,amount 共 5 行）再测试
class CSVReader:
    def __init__(self, path):
        try:
            with open(path, encoding='utf-8') as f:
                lines = [line.strip() for line in f if line.strip()]
        except FileNotFoundError as e:
            raise FileNotFoundError(f'文件不存在: {path}') from e

        if not lines:
            raise ValueError(f'文件为空: {path}')

        self.headers = lines[0].split(',')          # 第一行表头
        self.rows = [line.split(',') for line in lines[1:]]  # 其余是数据

    def column(self, name):
        if name not in self.headers:
            raise KeyError(f'列名不存在: {name}')
        idx = self.headers.index(name)
        return [row[idx] for row in self.rows]

    def mean(self, name):
        values = self.column(name)                  # 列名错会在这里抛 KeyError
        if not values:
            raise ValueError(f'列 {name} 没有数据')
        try:
            nums = [float(v) for v in values]
        except ValueError:
            raise ValueError(f'列 {name} 含有非数值数据: {values}')
        return sum(nums) / len(nums)
```
