## 4. 列表与切片

```python
nums = list(range(10))
print(nums[2:5], nums[:3], nums[-2:], nums[::2], nums[::-1])
nums.append(10); nums += [11, 12]
print(nums)
```

```python
letters = list('abcdefgh')   # ['a','b','c','d','e','f','g','h']

# 练习 A：取出 ['c','d','e']

print(letters[2:5])

# 练习 B：取出最后 3 个

print(letters[-3:])

# 练习 C：取出所有奇数位（b, d, f, h）——提示：想想起点和步长

print(letters[1::2])

# 练习 D：取出 ['a','c','e'] 三个

print(letters[:5:2])

# 练习 E：不反转列表，只取出「从第 6 个元素往回数 3 个」→ ['f','e','d']（难题，用负索引+负步长）

print(letters[5:2:-1])

# 练习 F：一句话取出列表的「中间两个元素」(d, e)

print(letters[len(letters)//2-1:len(letters)//2+1])
```

## 6. 晚间练习（全部自己写）

```python
# 练习 1：FizzBuzz —— 1~30，3 的倍数印 Fizz，5 的倍数印 Buzz，既是 3 又是 5 印 FizzBuzz

for i in range(1,31):
    if i % 15 == 0:
        print("FizzBuzz")
    elif i % 3 == 0:
        print("Fizz")
    elif i % 5 == 0:
        print("Buzz")
    else:
        print(i)

# 练习 2：反转字符串（先切片法，再 reversed() 法）

s = 'akrnwnf'
print(s[::-1])
print(''.join(reversed(s)))


# 练习 3：统计一段英文里每个单词出现的次数（用 dict）

text = 'the cat sat on the mat the cat'
words = text.split()
counts = {}
for w in words:
    counts[w] = counts.get(w,0) + 1
print(counts)

# 练习 4：找出列表中第二大且不重复的数

def second_largest(nums):
    unique = sorted(set(nums))
    if len(unique) < 2:
        return None          # 不存在，明确告诉调用者
    return unique[-2]



# 练习 5：判断一个数是否为质数（写函数 is_prime(n)）

def is_prime(n):
    if n < 2:
        return False
    if n == 2:
        return True
    if n % 2 == 0:
        return False
    i = 3
    while i * i <= n:      # 只试到 sqrt(n)
        if n % i == 0:
            return False
        i += 2
    return True


# 测试
for x in [1, 2, 3, 4, 5, 9, 11, 17, 25, 29]:
    print(x, is_prime(x))

# 练习 6：斐波那契前 30 项，放进列表

fib = [0, 1]

while len(fib) < 30:
    fib.append(fib[-1] + fib[-2])

print(fib)
print(len(fib))   # 30
```
