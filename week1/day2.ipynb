## 2. 推导式：同一逻辑写两遍，对比

```python
sq_loop = []
for i in range(10):
    if i % 2 == 0:
        sq_loop.append(i * i)
sq = [i * i for i in range(10) if i % 2 == 0]
print(sq_loop == sq)
print(sq)
# 再来：字典推导 {单词: 长度}；集合推导 {x % 5 for x in range(100)}
```

```python
# A：用推导式一行写出「1~20 里 3 的倍数加 1」→ [4,7,10,13,16,19]

res1 = [i + 1 for i in range(1,21) if i % 3 == 0]
print(res1)

# B：把 ['Alice','bob','CARL'] 一行规范成 ['alice','bob','carl']

words = ['Alice','bob','CARL']
res2 = [w.lower() for w in words]
print(res2)

# C：字典推导，从 dict {'a':1,'b':2,'c':3} 造出 {'a':1,'c':9}（只要值>1... 等等，值>1的键并平方值）
#   题目认真版：造 {键: 值的平方}，但只保留值>1的项 → {'b': 4, 'c': 9}

d = {'a':1,'b':2,'c':3}
res3 = {key: value * value for key, value in d.items() if value > 1}
print(res3)

# D：集合推导：{x*x for x in [-3..3]}（用 range(-3,4)），观察有几个元素、为什么

res4 = {x*x for x in range(-3,4)}
print(res4)

# E：嵌套：用嵌套推导式把 [[1,2],[3,4],[5,6]] 压平成一维 [1,2,3,4,5,6]（提示：外层遍历行，内层遍历元素，这次外层在前写）
matrix = [[1,2],[3,4],[5,6]]
res5 = [x for row in matrix for x in row]
print(res5)
```

## 3. 解包与 *args / **kwargs

```python
def profile(name, age, *hobbies, **extra):
    print(name, age, hobbies, extra)
profile('x', 20, 'code', 'run', city='BJ')
a, *rest = [1, 2, 3, 4]
print(a, rest)
```

```python
# A：def show(*args, **kwargs): print(args, kwargs)
#    调用 show(1, 2, x=3)，猜输出长什么样再跑

def show(*args, **kwargs): print(args, kwargs)
show(1, 2, x=3)

# B：不用 *args，用 * 解包实现：拿 [10, 20, 30, 40, 50] 的 head 和最后一个元素 tail，中间不要
#    print(head, tail)  # 应输出 10 50

nums = [10, 20, 30, 40, 50]
head, *mid, tail = nums
print(head, tail)

# C：def mean(*nums) 上面写过了——补上空参数保护：mean() 不崩，返回 0
#    （想想 len(nums) 是 0 时哪行会炸）

def mean(*nums):
    return sum(nums) / len(nums) if nums else 0
print(mean())        # 0
print(mean(1,2,3))   # 2.0


# D：现有函数 greet(name, greeting) 和字典 d = {'name': 'Bob', 'greeting': 'Hi'}
#    不许改成 greet(d['name'], d['greeting'])，用 ** 一行调用它
def greet(name, greeting):
    print(f"{greeting}, {name}!")


d = {'name': 'Bob', 'greeting': 'Hi'}

greet(**d)
```

## 5. 练习

```python
# 练习 1：统计文本词频，输出出现次数最多的前 10 个词（sorted + key=lambda）
text = 'the quick brown fox jumps over the lazy dog the fox runs'

words = text.split()
d = {}
for w in words:
    d[w] = d.get(w,0) + 1

top10 = sorted(d.items(),key = lambda t: -t[1])[:10]
print(top10)

# 练习 2：列表去重并保持原有顺序
nums = [3, 1, 4, 1, 5, 9, 2, 6, 5, 3]
res = list(dict.fromkeys(nums))
print(res)   # [3, 1, 4, 5, 9, 2, 6]

nested = {'a': {'b': 1, 'c': 2}, 'x': {'y': 9}}

res = {}
for outer_key, inner in nested.items():
    for inner_key, value in inner.items():
        res[f'{outer_key}.{inner_key}'] = value

print(res)
# {'a.b': 1, 'a.c': 2, 'x.y': 9}

# 练习 4：用嵌套推导式转置矩阵 [[1,2,3],[4,5,6]] -> [[1,4],[2,5],[3,6]]

matrix = [[1,2,3],[4,5,6]]
res = [[row[i] for row in matrix] for i in range(len(matrix[0]))]
print(res) 
```

## 6. 晚间项目：JSON 通讯录
要求：add / find / delete / list 四个函数；数据存 contacts.json；程序重启后数据还在

```python
import json
# 在这里实现
import json
import os

FILE = 'contacts.json'


def load():
    """读取通讯录。文件不存在（第一次运行）返回空字典。"""
    if not os.path.exists(FILE):
        return {}
    with open(FILE, 'r', encoding='utf-8') as f:
        return json.load(f)


def save(data):
    """把字典写回 JSON 文件。"""
    with open(FILE, 'w', encoding='utf-8') as f:
        json.dump(data, f, ensure_ascii=False, indent=2)


def add(name, phone):
    """添加联系人。同名已存在返回 False，不覆盖；成功返回 True。"""
    data = load()
    if name in data:
        return False
    data[name] = phone
    save(data)
    return True


def find(name):
    """查找号码。查不到返回 None，不崩。"""
    data = load()
    return data.get(name)


def delete(name):
    """删除联系人。查不到返回 False；删掉并保存返回 True。"""
    data = load()
    if name not in data:
        return False
    del data[name]
    save(data)
    return True


def list_all():
    """打印所有联系人。"""
    data = load()
    if not data:
        print('通讯录为空')
        return
    for name, phone in data.items():
        print(f'{name}: {phone}')


if __name__ == '__main__':
    print(add('张三', '13800000000'))   # True
    print(add('张三', '13900000000'))   # False，不覆盖
    print(add('李四', '13700000000'))   # True
    print(find('张三'))                 # 13800000000
    print(find('王五'))                 # None
    print(delete('李四'))               # True
    print(delete('王五'))               # False
    list_all()
```
